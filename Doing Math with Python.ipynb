```python
'''
Unit converter: Miles and Kilometers
'''
def print_menu():
    print('1. Kilometers to Miles')
    print('2. Miles to Kilometers')
def km_miles():
    km = float(input('Enter distance in kilometers:'))
    miles = km / 1.609
    print('Distance in miles: {0}'.format(miles))
def miles_km():
    miles = float(input('Enter distance in miles: '))
    km =miles * 1.609
    print('Distance in kilometers: {0}'.format(km))
if __name__ == '__main__':
    print_menu()
    choice = input('Which conversion would you like to do?:')
    if choice == '1':
        km_miles()
    if choice == '2':
        miles_km()

print_menu()
km_miles()
```

```python
'''
Quadratic equation root calculator
'''
def roots(a, b, c):
    D = (b*b - 4*a*c)**0.5
    x_1 = (-b + D)/(2*a)
    x_2 = (-b - D)/(2*a) 
    
    print('x1: {0}'.format(x_1))
    print('x2: {0}'.format(x_2))
if __name__ == '__main__':
    a = input('Enter a: ')
    b = input('Enter b: ')
    c = input('Enter c: ')
    roots(float(a), float(b), float(c))

roots (10, 15, 87)
```

대괄호 안에 쉼표로 구분된 값을 입력하여 목록을 만들 수 있습니다. 다음 문장은 목록을 생성하고 레이블 simplelist 를 사용하여 이를 참조합니다

```python
simplist = [1,2,3]
```

이제 레이블과 목록에서 해당 숫자의 위치를 사용하여 1, 2, 3이라는 개별 숫자를 참조할 수 있으며, 이를 인덱스라고 합니다. 따라서 simplelist[0]는 첫 번째 숫자를 의미하고, simplelist[1]은 두 번째 숫자를 의미하며, simplelist[2]는 세 번째 숫자를 의미합니다:

```python
simplist [0] 
simplist [1] 
simplist [2]
```

목록의 첫 번째 항목이 인덱스 0에 있고, 두 번째 항목이 인덱스 1에 위치한다는 점에 유의하십시오. 즉, 목록의 위치는 1이 아니라 0부터 계산됩니다.

리스트는 문자열도 저장할 수 있습니다:

```python
stringlist = ['a string','b string','c string']
stringlist[0] # 'a string'
stringlist[1] # 'b string'
stringlist[2] # 'c string'
```

리스트를 만드는 한 가지 장점은 각 값마다 별도의 라벨을 만들 필요가 없다는 점이며, 리스트에 라벨을 만들고 인덱스 위치를 사용하여 각 항목을 참조하면 됩니다. 또한, 새로운 값을 저장해야 할 때마다 목록에 추가할 수 있으므로, 미리 몇 개의 숫자나 문자열을 저장해야 할지 모를 경우 데이터를 저장하기 위한 목록이 가장 좋은 선택입니다.

빈 목록은 바로 그—항목이나 요소가 전혀 없는 목록—이며, 다음과 같이 만들 수 있습니다:

```python
emptylist = []
```

빈 목록은 사전에 목록에 포함될 항목을 알 수 없지만 프로그램 실행 중에 값을 채울 계획일 때 주로 유용합니다. 그 경우에는 빈 리스트를 만든 다음, 나중에 append() 메서드를 사용하여 항목을 추가할 수 있습니다

```python
emptylist.append(1) 
emptylist.append('korea')
emptylist
```

➊에서 빈 목록은 빈 상태로 시작합니다. 다음으로, 
➋에 숫자 1을 추가한 다음 
➌에 2를 추가합니다. 
줄 ➍에 따라, 이제 목록은 [1, 2]입니다. 
.Append()를 사용할 때 값이 목록의 끝에 추가된다는 점에 유의하십시오. 이는 리스트에 값을 추가하는 한 가지 방법일 뿐입니다. 다른 것이 있지만, 이번 장에서는 필요하지 않을 것입니다.

튜플을 만드는 것은 목록을 만드는 것과 유사하지만, 대괄호 대신 괄호를 사용합니다:

```python
simpletuple = (1,2,"myTuple")
simpletuple
```

리스트와 마찬가지로, 괄호 안의 해당 인덱스를 사용하여 단순수에서 개별 숫자를 참조할 수 있습니다.

```python
simpletuple[0]
simpletuple[1]
simpletuple[2]
```

리스트와 튜플 모두에 대해 음수 인덱스를 사용할 수도 있습니다. 예를 들어, simplelist[-1]와 simpletuple[-1]은 리스트 또는 튜플의 마지막 요소를 가리키며, simplelist[-2]와 simpletuple[-2]는 두 번째 끝 요소를 가리키고, 이와 같이 계속됩니다.

튜플은 리스트와 마찬가지로 문자열을 값으로 가질 수 있으며, emptytuple=()와 같이 요소가 없는 빈 튜플을 만들 수 있습니다. 하지만 기존 튜플에 새 값을 추가하는 append() 메서드가 없기 때문에 빈 튜플에 값을 추가할 수 없습니다. 튜플을 생성하면 튜플의 내용을 변경할 수 없습니다.
